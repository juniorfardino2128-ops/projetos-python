```python
# bebida
quntde_coca = 500
preço_coca = 3.50
quntde_pespi = 800
preço_pespi = 4.00
quntde_fanta = 350
preço_fanta = 2.50
quntde_doly = 300
preço_doly = 2.80
custo = 4000
```

```python
# venda coca
print(quntde_coca * preço_coca)
```

```python
# venda pespi
print(quntde_pespi * preço_pespi)
```

```python
# venda fanta
print(quntde_fanta * preço_fanta)
```

```python
# venda doly
print(quntde_doly * preço_doly)
```

```python
# descobrindo faturamento da fabrica
faturamento = quntde_coca * preço_coca + quntde_pespi * preço_pespi + quntde_fanta * preço_fanta + quntde_doly * preço_pespi
print(faturamento)
```

```python
# descobrindo o lucro da fabrica
lucro = faturamento - custo
print('O lucro da fabrica Coca Cola foi de {}'.format(lucro))
```
