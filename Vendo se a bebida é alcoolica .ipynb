```python
def ehalcoolico(bebida):
    bebida = bebida.upper()
    if 'BEB' in bebida :
        return True
    else:
        return False
    
    
produtos = ['beb2000', 'TAF45602', 'CAC15020', 'BEB89745', 'BEB56321', 'ded12045', 'beb36923', 'TAF48265', 'BEB75224', 'CAC432525', 'beb15054','TAF40020', 'BEB89520']
for produto in produtos:
    if ehalcoolico(produto):
        print('Envier essa {}, mensagem para setor de bebida alcoolica'.format(produto))
```
