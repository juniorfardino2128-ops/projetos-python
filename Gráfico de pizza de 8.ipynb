```python
import matplotlib.pyplot as plt

valores = [80, 70, 60, 58, 40, 38, 35,30]
produtos = ['geladeira', 'sofá', 'fogão', 'microonda', 'ferro de passar', 'TV', 'guarda-roupa', 'cama']
```

```python
rotulos = [f'{produto}: {quantidade}' for produto, quantidade in zip(produtos, valores)]
import matplotlib.pyplot as plt
from cycler import cycler
cores = plt.get_cmap('Set1').colors
ciclo_cores = cycler('color', cores)
plt.rc('axes', prop_cycle=ciclo_cores)

destaque = [0.1, 0.1, 0, 0, 0, 0, 0, 0]

plt.pie(valores, labels=produtos, autopct='%.1f%%', explode=destaque, shadow=True)
plt.title('Vendo vendas da loja')
plt.show()
```
